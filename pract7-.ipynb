Q.1 Linear Regression (House Price Prediction)

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = {
    "Area": [600, 800, 1000, 1200, 1400, 1600],
    "Price": [32, 40, 49, 58, 68, 79]
}

df = pd.DataFrame(data)

print(df)

# Independent and Dependent variables
X = df[["Area"]]
y = df["Price"]

# Train model
model = LinearRegression()
model.fit(X, y)

# Slope and Intercept
m = model.coef_[0]
c = model.intercept_

print("\nSlope =", m)
print("Intercept =", c)

print("\nRegression Equation:")
print(f"Price = {m:.4f} × Area + {c:.4f}")

# Predictions
price1500 = model.predict([[1500]])
price2500 = model.predict([[2500]])

print("\nPredicted Price for 1500 sq.ft =", round(price1500[0],2), "Lakhs")
print("Predicted Price for 2500 sq.ft =", round(price2500[0],2), "Lakhs")
```

Q.2 Employee Salary Prediction using Linear Regression

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Create DataFrame
data = {
    "Employee ID":[1,2,3,4,5,6,7,8],
    "Age":[22,24,26,28,30,32,34,36],
    "Experience":[1,2,3,4,5,6,8,10],
    "Salary":[25000,30000,35000,40000,45000,50000,60000,70000]
}
df = pd.DataFrame(data)
print(df)

# Independent and Dependent variables
X = df[["Experience"]]
Y = df["Salary"]

# Split dataset
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# Train model
model = LinearRegression()
model.fit(X_train, Y_train)

# Predict salary for 7 years experience
salary = model.predict([[7]])
print("\nPredicted Salary for 7 Years Experience =", salary[0])

# Testing prediction
Y_pred = model.predict(X_test)

# Display results
print("\nSlope (Coefficient) =", model.coef_[0])
print("Intercept =", model.intercept_)
print("R2 Score =", r2_score(Y_test, Y_pred))

# Plot Regression Line
plt.scatter(X, Y)
plt.plot(X, model.predict(X))
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
print("Aashu")
plt.show()
```
