```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("train.csv")

# Display first 5 rows
print(df.head())

# Dataset shape
print("Shape of dataset:", df.shape)
```

```python
print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nColumn Names:")
print(df.columns)
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
target = "Survived"

print("Target Variable:", target)
print(df[target].value_counts())
```

```python
print(df["Survived"].value_counts())

print("\nPercentage Distribution:")
print(df["Survived"].value_counts(normalize=True) * 100)
```

```python
sns.countplot(x="Survived", data=df)

plt.title("Survival Class Distribution")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
print(df["Sex"].value_counts())

sns.countplot(x="Sex", data=df)

plt.title("Gender Representation")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
gender_survival = pd.crosstab(df["Sex"], df["Survived"])

print(gender_survival)

sns.countplot(x="Sex", hue="Survived", data=df)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

```python
print(df["Pclass"].value_counts().sort_index())

sns.countplot(x="Pclass", data=df)

plt.title("Passenger Class Representation")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
print(pd.crosstab(df["Pclass"], df["Survived"]))

sns.countplot(x="Pclass", hue="Survived", data=df)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)

print(df["AgeGroup"].value_counts())
```

```python
sns.countplot(x="AgeGroup", data=df)

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=30)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("german_credit_data.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
target = "Risk"

print("Target Variable:", target)

print("\nTarget Distribution:")
print(df["Risk"].value_counts())
```

```python
print(df["Risk"].value_counts())

print("\nPercentage:")
print(df["Risk"].value_counts(normalize=True) * 100)
```

```python
sns.countplot(x="Risk", data=df)

plt.title("Credit Risk Class Distribution")
plt.xlabel("Credit Risk")
plt.ylabel("Number of Customers")
plt.show()
```

```python
print(pd.crosstab(df["Sex"], df["Risk"]))

sns.countplot(x="Sex", hue="Risk", data=df)

plt.title("Credit Risk by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 50, 65, 100],
    labels=["18-25", "26-35", "36-50", "51-65", "66+"]
)

print(df["AgeGroup"].value_counts().sort_index())

sns.countplot(x="AgeGroup", data=df)

plt.title("Age Group Representation")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.show()
```

```python
print(pd.crosstab(df["AgeGroup"], df["Risk"]))

sns.countplot(x="AgeGroup", hue="Risk", data=df)

plt.title("Credit Risk by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.legend(title="Risk")
plt.show()
```

```python
print(pd.crosstab(df["Housing"], df["Risk"]))

sns.countplot(x="Housing", hue="Risk", data=df)

plt.title("Credit Risk by Housing Category")
plt.xlabel("Housing Category")
plt.ylabel("Number of Customers")
plt.legend(title="Risk")
plt.show()
```
